```python
# ============================================================
# STEP 1: Load the dataset after OUTLIER REMOVAL
# ============================================================

import pandas as pd

dataset = pd.read_csv("/kaggle/input/notebooks/alkaraikbar/outlier-finalversion")

print("Dataset loaded successfully!")
print("Dataset shape:", dataset.shape)


# ============================================================
# STEP 2: Check for missing values
# ============================================================

# Missing values can cause problems during model training.

print("\nMissing values:")
print(dataset.isnull().sum())


# ============================================================
# STEP 3: Check for duplicate rows
# ============================================================

# Duplicate patient records can unnecessarily influence the model.

duplicate_rows = dataset.duplicated().sum()

print("\nDuplicate rows:", duplicate_rows)

# Remove duplicate rows if they exist.

dataset = dataset.drop_duplicates().copy()

print("Shape after duplicate removal:", dataset.shape)


# ============================================================
# STEP 4: Check for duplicate patient IDs
# ============================================================

# Each patient should have a unique ID.

duplicate_ids = dataset["id"].duplicated().sum()

print("\nDuplicate patient IDs:", duplicate_ids)


# ============================================================
# STEP 5: Validate binary features
# ============================================================

# These features should contain only 0 and 1.

binary_features = [
    "smoke",
    "alco",
    "active",
    "cardio"
]

for feature in binary_features:
    print(
        feature,
        "unique values:",
        sorted(dataset[feature].unique())
    )


# ============================================================
# STEP 6: Validate ordinal features
# ============================================================

# cholesterol:
# 1 = normal
# 2 = above normal
# 3 = well above normal
#
# gluc:
# 1 = normal
# 2 = above normal
# 3 = well above normal

print("\nCholesterol values:",
      sorted(dataset["cholesterol"].unique()))

print("Glucose values:",
      sorted(dataset["gluc"].unique()))


# ============================================================
# STEP 7: Check for invalid categorical values
# ============================================================

invalid_cholesterol = dataset[
    ~dataset["cholesterol"].isin([1, 2, 3])
]

invalid_gluc = dataset[
    ~dataset["gluc"].isin([1, 2, 3])
]

print("\nInvalid cholesterol rows:",
      invalid_cholesterol.shape[0])

print("Invalid glucose rows:",
      invalid_gluc.shape[0])


# ============================================================
# STEP 8: Check data types
# ============================================================

# Verify that all features have appropriate numerical types.

print("\nData types:")
print(dataset.dtypes)


# ============================================================
# STEP 9: Final missing-value check
# ============================================================

print("\nFinal missing-value check:")
print(dataset.isnull().sum())


# ============================================================
# STEP 10: Save FINAL cleaned dataset
# ============================================================

# This dataset will be used as input for the next stage:
# Feature Engineering / Model Preparation.

dataset.to_csv(
    "../data/processed/cardio_final_cleaned.csv",
    index=False
)

print("\nFinal cleaned dataset saved successfully!")
print("Final dataset shape:", dataset.shape)
```
